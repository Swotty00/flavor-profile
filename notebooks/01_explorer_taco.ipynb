```python
import pandas as pd
from pathlib import Path

path = Path("..")/"data"/"raw"/"taco.xlsx"
pd.set_option("display.max_columns", None)

with pd.ExcelFile(path) as xlsx:
    df_raw = pd.read_excel(xlsx, sheet_name=0, header=None, keep_default_na=False)
    df1 = pd.read_excel(xlsx, sheet_name=1, header=None, keep_default_na=False)
    df2 = pd.read_excel(xlsx, sheet_name=2, header=None, keep_default_na=False)

df_nomeado = df_raw.copy()
```

```python
print(len(df_nomeado.columns))
header = ['id', 'descricao', 'umidade_pct', 'energia_kcal', 'energia_kj', 'proteina_g', 'lipideo_g', 'colesterol_mg', 'carboidrato_g', 'fibra_g', 'cinzas_g', 'calcio_mg', 'magnesio_mg', 'id_repetido', 'manganes_mg', 'fosforo_mg', 'ferro_mg', 'sodio_mg', 'potassio_mg', 'cobre_mg', 'zinco_mg', 'retinol_mcg', 're_mcg', 'rae_mcg', 'tiamina_mg', 'riboflavina_mg', 'piridoxina_mg', 'niacina_mg', 'vitaminac_mg']

assert len(header) == len(df_nomeado.columns), "O tamanho do cabeçalho é diferente do número de colunas"
df_nomeado.columns = header
```

```python
# Corta a o cabeçalho e a legenda
df_cortado = df_nomeado.iloc[3:687, :].reset_index(drop=True)
df_cortado.head()
```

```python
import numpy as np

is_food = pd.to_numeric(df_cortado['id'], errors='coerce').notna()
isnt_food = df_cortado[~is_food]['id'].value_counts()

display(isnt_food)

garbage = ['', 'Número do', 'Alimento']

df_cat = df_cortado.copy()

df_cat['categoria'] = (
    df_cat['id']
    .where(~is_food)
    .replace(garbage, np.nan)
    .ffill()
)

df_alimentos = df_cat[is_food].copy()

df_alimentos['id'] = df_alimentos['id'].astype(int)
df_alimentos = df_alimentos.reset_index(drop=True)

display(df_alimentos.head())
print(f"\nContagem por categoria: {df_alimentos['categoria'].value_counts()}")
```

```python
temporary = pd.to_numeric(df_alimentos['id_repetido'], errors='coerce')
rep = (df_alimentos['id'] == temporary).all()

assert rep, "Existem IDs diferentes"

df_sem_dup = df_alimentos.drop('id_repetido', axis=1)
```

```python
for column in df_sem_dup.columns:
    if column not in ['id', 'descricao', 'categoria']:
        temporary = pd.to_numeric(df_sem_dup[column], errors='coerce')
        mask = temporary.isna()
        print(f"Contagem da {column}:\n{df_sem_dup[column].where(mask).value_counts()}")
```

```python
mask_um = df_sem_dup['umidade_pct'] == 'NA'
mask_prot = df_sem_dup['proteina_g'] == 'NA'

caract = ['id','descricao','categoria','energia_kcal']

df_sem_dup.loc[mask_um, caract]
df_sem_dup.loc[mask_prot, caract]

# basicamente estou averiguando o que 'NA', '', 'Tr" e '*' significam, levando em conta os nutrientes
# reais dos alimentos.
# 'NA': 0

nutrientes = []

for column in df_sem_dup.columns:
    if column not in ['id', 'descricao', 'categoria']:
        nutrientes.append(column)

mask_ast = df_sem_dup['umidade_pct'] == '*'
df_sem_dup.loc[mask_ast, caract]
```

```python
pedaco = df_sem_dup.loc[mask_ast, nutrientes]
contagem_ast = (pedaco == '*').sum(axis=1)
display(contagem_ast)
len(nutrientes)
```

```python
mark = set()
for column in nutrientes:
    col = df_sem_dup[column]
    falhou = pd.to_numeric(col, errors='coerce').isna()
    marcadores = col[falhou].unique()
    mark.update(marcadores)

display(mark)

# surgem alguns erros de digitação que devem ser tratados, como ',0,02' e outras variações
```

```python
import numpy as np

switch = [' Tr', 'Tr', 'NA']
dic = dict.fromkeys(mark, np.nan)
dic[',0,02'] = 0.02
dic.update(dict.fromkeys(switch, 0))

df_sem_dup[nutrientes] = (df_sem_dup[nutrientes].replace(dic)).astype(float)
```
